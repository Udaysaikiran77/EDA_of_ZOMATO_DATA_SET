```python
# second dataset
df_country = pd.read_excel("Country-Code.xlsx")
df_country.head()
```

```python
# to combine two data_sets
#pd.merge(leftdataset, rightdataset, on = which column, how = to merge on which table left or right
final_df =pd.merge(df,df_country,on='Country Code',how = 'left')
```

```python
## find out how many values are there based on those features
country_name = final_df.Country.value_counts().index
# this gives how many values are listed on that paticula name or word or records in afeature
```

```python
# grouping multple features as a single dataframe
ratings = final_df.groupby(['Aggregate rating','Rating color', 'Rating text']).size().reset_index().rename(columns={0:'rating count'})
```

```python
ratings.head()
```

```python
# observation: 
# 1. when ever the rating is between 4.5 to 4.9 >>> it shows dark green colour and excelllent 
# 2. when ever the rating is between 4.0 to 4.4 >>> its represnts green colour and its rated as very good
# 3. when ever the rating is between 3.5 to 3.9 >>> its shows that good and it represents in yellow
# 4. when ever the rating is between 2.5 to 3.4 >>> its shows that Average  and it represents in orange 
# 5. when ever the rating is between 1.8 to 2.4 >>> its shows that poor  and it represents in red
# 6. and among the all the ratingis given people there more people who have not given rating and the count is 2148
```

```python
# ploting a barplot using seaborn , to check relation between the two features 
#matplotlib.rcParams['figure.figsize'] = (12,6) this command is used to adujust the size of the barplot
import matplotlib
matplotlib.rcParams['figure.figsize'] = (12,6)
sns.barplot(x='Aggregate rating',y='rating count', data = ratings)
```

```python
# hue is used for setting the colurs, for macthing or for using  the excat color to the color which is in the dataframe we use palette =   ['white','red','orange','yellow','green','dark green']
sns.barplot(x='Aggregate rating',y='rating count',hue='Rating color', data = ratings, palette = ['white','red','orange','yellow','green','green'])
```

```python
# obervations;
# 1. not rated count is high compartivly
# 2. the max number of rating is between 2.5 to 3.5
```

```python
# ploting the count based on how many records are baseed on that paticular color
sns.countplot(x = 'Rating color', data = ratings, palette = ['white','red','orange','yellow','green','green'])
```
